```python
#x = np.array([1,2,3,4,5,1,2,1,1,1])  
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

# Find unique values and their counts
values, counts = np.unique(x, return_counts=True)

# Find the value with the highest frequency
most_frequent = values[np.argmax(counts)]

# Find the indices of the most frequent value
indices = np.where(x == most_frequent)[0]

print("Array x:")
print(x)

print("Most frequent value:", most_frequent)
print("Number of times it occurs:", counts.max())
print("Indices:", indices)
```

```python
import numpy as np

# Create a 1-D NumPy array with 25 evenly spaced numbers
mehakdeep_kaur = np.linspace(10, 100, 25)

print("1-D NumPy Array:")
print(mehakdeep_kaur)

# Print dimensions
print("\nDimensions of array:", mehakdeep_kaur.ndim)

# Print shape
print("Shape of array:", mehakdeep_kaur.shape)

# Print total number of elements
print("Total number of elements:", mehakdeep_kaur.size)

# Print data type
print("Data type of elements:", mehakdeep_kaur.dtype)

# Print total bytes consumed
print("Total bytes consumed:", mehakdeep_kaur.nbytes)
```

```python
# Convert 1-D array into a 25 x 1 array
reshaped_array = mehakdeep_kaur.reshape(25, 1)

print("After reshape(25, 1):")
print(reshaped_array)

# Find transpose
transpose_array = reshaped_array.T

print("\nTranspose of the array:")
print(transpose_array)

print("\nShape after transpose:", transpose_array.shape)
```

```python
# Using T directly on the original 1-D array
transpose_using_T = mehakdeep_kaur.T

print("Transpose using T attribute:")
print(transpose_using_T)

print("\nShape using T:", transpose_using_T.shape)
```

```python
import numpy as np

# Create a 2-D NumPy array with 3 rows and 4 columns
ucs420_mehakdeep_kaur = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original 2-D Array:")
print(ucs420_mehakdeep_kaur)

# Calculate mean
print("\nMean:")
print(np.mean(ucs420_mehakdeep_kaur))

# Calculate median
print("\nMedian:")
print(np.median(ucs420_mehakdeep_kaur))

# Find maximum value
print("\nMaximum:")
print(np.max(ucs420_mehakdeep_kaur))

# Find minimum value
print("\nMinimum:")
print(np.min(ucs420_mehakdeep_kaur))

# Find unique elements
print("\nUnique Elements:")
print(np.unique(ucs420_mehakdeep_kaur))

# Reshape the array to 4 rows and 3 columns
reshaped_ucs420_mehakdeep_kaur = ucs420_mehakdeep_kaur.reshape(4, 3)

print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_mehakdeep_kaur)
```
