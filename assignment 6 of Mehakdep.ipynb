Q1. Plot Two Mathematical Functions
Question

Ask the user to enter a value M for a mathematical function.

Generate x values from -10 to 10 using np.linspace().
Compute:
y = M × x²
y = M × sin(x)
Plot both functions in a single figure.
Use different colors and line styles.
Add legend, grid and title.

```python
import numpy as np
import matplotlib.pyplot as plt

# Ask the user to enter the value of M
M = float(input("Enter the value of M: "))

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 200)

# Calculate y values
y1 = M * x**2
y2 = M * np.sin(x)

# Plot both functions
plt.figure(figsize=(10, 6))

plt.plot(x, y1, label="y = Mx²", linestyle='-', color='blue')
plt.plot(x, y2, label="y = Msin(x)", linestyle='--', color='red')

# Add title and labels
plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")

# Add legend and grid
plt.legend()
plt.grid(True)

plt.show()
```

Q2. Seaborn Bar Plot of Five Subjects and Scores
Question

Create a dataset of five subjects and scores.

Convert it into a Pandas DataFrame.
Plot scores using a Seaborn bar plot.
Use different colors for each bar.
Add annotations on top of each bar.
Add title, axis labels and grid.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create dataset
data = {
    "Subject": ["Python", "DBMS", "DAA", "Computer Architecture", "Machine Learning"],
    "Score": [85, 78, 88, 82, 91]
}

# Convert data into DataFrame
df = pd.DataFrame(data)

print("Dataset:")
print(df)

# Set figure size
plt.figure(figsize=(10, 6))

# Create Seaborn bar plot
ax = sns.barplot(
    data=df,
    x="Subject",
    y="Score",
    hue="Subject",
    legend=False,
    palette="Set2"
)

# Add annotations on top of each bar
for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3)

# Add title and labels
plt.title("Scores in Different Subjects")
plt.xlabel("Subjects")
plt.ylabel("Scores")

# Add grid
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.xticks(rotation=20)
plt.show()
```


#Q3. Generate Random Dataset and Create Subplots
Question

Use your roll number as the NumPy random seed and generate a dataset of 50 values using np.random.randn().#
# rollno = 1024170364#

```python
import numpy as np
import matplotlib.pyplot as plt

# Set roll number as random seed
np.random.seed(1024170364)

# Generate 50 random values
data = np.random.randn(50)

# Create x values
x = np.arange(1, 51)

# Calculate cumulative sum
cumulative_sum = np.cumsum(data)

# Create random noise
noise = np.random.randn(50)

# Create 2 x 2 subplot layout
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# -----------------------------
# Plot 1: Original Random Data
# -----------------------------
axes[0, 0].plot(x, data)
axes[0, 0].set_title("Random Data")
axes[0, 0].set_xlabel("Index")
axes[0, 0].set_ylabel("Value")
axes[0, 0].grid(True)

# -----------------------------
# Plot 2: Cumulative Sum
# -----------------------------
axes[0, 1].plot(x, cumulative_sum)
axes[0, 1].set_title("Cumulative Sum")
axes[0, 1].set_xlabel("Index")
axes[0, 1].set_ylabel("Cumulative Value")
axes[0, 1].grid(True)

# -----------------------------
# Plot 3: Scatter Plot
# -----------------------------
axes[1, 0].scatter(x, data)
axes[1, 0].set_title("Scatter Plot")
axes[1, 0].set_xlabel("Index")
axes[1, 0].set_ylabel("Value")
axes[1, 0].grid(True)

# -----------------------------
# Plot 4: Random Noise
# -----------------------------
axes[1, 1].scatter(x, noise)
axes[1, 1].set_title("Scatter Plot with Random Noise")
axes[1, 1].set_xlabel("Index")
axes[1, 1].set_ylabel("Noise")
axes[1, 1].grid(True)

# Adjust spacing
plt.tight_layout()

plt.show()
```

Q4(2). Display All Product Sales Data Using Multiline Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Product columns
products = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer"
]

# Create the figure
plt.figure(figsize=(12, 7))

# Plot sales of each product
for product in products:
    sns.lineplot(
        data=df,
        x="month_number",
        y=product,
        marker="o",
        label=product
    )

# Add title and labels
plt.title("Monthly Sales of All Products")
plt.xlabel("Month Number")
plt.ylabel("Sales Units")

# Add legend and grid
plt.legend()
plt.grid(True)

plt.show()
```

Q4(3). Bar Chart for All Features/Attributes

```python
# Product columns
products = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer"
]

# Calculate total sales of each product
product_totals = df[products].sum()

# Display total sales
print("Total Sales of Each Product:")
print(product_totals)

# Create bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    x=product_totals.index,
    y=product_totals.values
)

# Add title and labels
plt.title("Total Sales of All Products")
plt.xlabel("Products")
plt.ylabel("Total Sales")

# Rotate product names for better visibility
plt.xticks(rotation=20)

# Add grid
plt.grid(axis="y")

plt.show()
```
