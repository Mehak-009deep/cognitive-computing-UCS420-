Q1. Lists

```python
# Roll Number
roll_no = "1024170364"

# Create list L by multiplying each digit by 10
L = [int(digit) * 10 for digit in roll_no]

# i. Print L
print("Original List L:")
print(L)
```

```python
# Append a number
L.append(50)
print("After append(50):")
print(L)

# append() adds the element at the end of the list.

# Insert a number at index 3
L.insert(3, 99)
print("After insert(3,99):")
print(L)

# insert() places the element at the specified position.
```

```python
# remove() removes the first occurrence of a value
L.remove(50)
print("After remove(50):")
print(L)

# pop() removes element at given index
L.pop(2)
print("After pop(2):")
print(L)
```

```python
# Ascending
L.sort()
print("Ascending Order:")
print(L)

# Descending
L.sort(reverse=True)
print("Descending Order:")
print(L)
```

```python
print("First 3 Elements:", L[:3])
print("Last 3 Elements:", L[-3:])
```

```python
average = sum(L) / len(L)

greater_than_avg = [x for x in L if x > average]

print("Average =", average)
print("Elements greater than average:")
print(greater_than_avg)
```

```python
# Use first 8 values from original list
L_original = [int(digit) * 10 for digit in roll_no]

scores = tuple(L_original[:8])

print(scores)
```

```python
highest = max(scores)
lowest = min(scores)

print("Highest Score:", highest)
print("Highest Index:", scores.index(highest))

print("Lowest Score:", lowest)
print("Lowest Count:", scores.count(lowest))
```

```python
# Tuples are immutable, so they cannot be modified in place.

reverse_list = list(scores[::-1])

print(reverse_list)
```

```python
value = int(input("Enter a score: "))

if value in scores:
    print("First occurrence index:", scores.index(value))
else:
    print("Score not present.")
```

```python
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)

# Tuples are immutable, whereas lists are mutable.
```

```python
first, second, *remaining = scores

print("First:", first)

print("Second:", second)
print("Remaining:", remaining)
```

```python
import random

random.seed(1024170364)

numbers = [random.randint(100,900) for _ in range(100)]

print(numbers)
```

```python
ii. Odd Numbers
```

```python
odd = [x for x in numbers if x % 2 == 1]

print("Odd Count:", len(odd))
print(odd)
```

```python
iii. Even Numbers
```

```python
even = [x for x in numbers if x % 2 == 0]

print("Even Count:", len(even))
print(even)
```

```python
iv. Prime Numbers
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False

    return True

prime_numbers = [x for x in numbers if is_prime(x)]

print("Prime Count:", len(prime_numbers))
print(prime_numbers)
```

```python
from collections import Counter

count = Counter(numbers)

most_common = count.most_common(1)[0]

print("Most Frequent Number:", most_common[0])
print("Frequency:", most_common[1])
```

```python
digits = [int(d) for d in roll_no[:8]]

A = {x*7 for x in digits}
B = {x*9 for x in digits}

print("Set A:", A)
print("Set B:", B)
```

```python
print("Union:")
print(A.union(B))
```

```python
print("Intersection:")
print(A.intersection(B))
```

```python
print("A - B:")
print(A.difference(B))

print("B - A:")
print(B.difference(A))

# difference() returns elements only in one set,
# whereas symmetric_difference() returns elements unique to both sets.
```

```python
print("Symmetric Difference:")
print(A.symmetric_difference(B))
```

```python
print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))
```

```python
X = int(input("Enter value to remove from Set A: "))

A.discard(X)

print("Updated Set A:")
print(A)

# discard() does not raise an error if the element is absent,
# whereas remove() raises a KeyError.
```

```python
# Q5: Dictionary

name = "Mehakdeep Kaur"
roll_no = "1024170364"
branch = "CSE"

age = int(input("Enter your age: "))
cgpa = float(input("Enter your CGPA: "))

my_dict = {
    "name": name,
    "roll_no": roll_no,
    "branch": branch,
    "age": age,
    "city": "Patiala"
}

print("Original Dictionary:")
print(my_dict)
```

```python
# Rename the key "city" to "location" without changing its value

my_dict["location"] = my_dict.pop("city")

print("After renaming city to location:")
print(my_dict)

# pop() removes "city" and returns its value,
# which is then assigned to the new key "location".
```

```python
# Add CGPA to the dictionary

my_dict["cgpa"] = cgpa

print("After adding CGPA:")
print(my_dict)
```

```python
# Increase age by 1

my_dict["age"] = my_dict["age"] + 1

print("After increasing age by 1:")
print(my_dict)
```

```python
# Create two copies of the dictionary

dict_pop = my_dict.copy()
dict_del = my_dict.copy()

# Method 1: Using pop()
removed_value = dict_pop.pop("branch")

print("Dictionary after using pop():")
print(dict_pop)
print("Value returned by pop():", removed_value)

# Method 2: Using del
del dict_del["branch"]

print("\nDictionary after using del:")
print(dict_del)

# pop() removes the key and returns its value,
# whereas del only removes the key-value pair and does not return the value.
```

```python
# Print every key-value pair using items()

print("Key → Value")

for key, value in my_dict.items():
    print(f"{key} → {value}")
```

```python
# Check whether email exists before accessing it

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email is not present in the dictionary.")
```

```python
# Create a fictional friend's dictionary
friend_dict = {
    "name": "Simran Kaur",
    "roll_no": "1024170370",
    "branch": "CSE",
    "age": 20,
    "city": "Chandigarh"
}

print("Friend Dictionary:")
print(friend_dict)

# Merge both dictionaries using ** syntax
merged_dict = {**my_dict, **friend_dict}

print("\nMerged Dictionary:")
print(merged_dict)

# When both dictionaries have the same key,
# the value from the second dictionary (friend_dict) wins.
```

```python
# Create a new dictionary containing only key-value pairs
# whose values are strings

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Dictionary containing only string values:")
print(string_values)
```
