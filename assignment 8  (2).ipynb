Q1. Text Preprocessing and Word Frequency Distribution

```python
text = """
Artificial intelligence is changing the world of technology.
It helps people solve problems and complete tasks faster.
Machine learning allows computers to learn from data.
I enjoy learning Python because it makes programming easier.
In the future, intelligent systems may improve education and healthcare.
"""

print("Original Paragraph:")
print(text)
```

```python
clean_text = re.sub(r'[^\w\s]', '', text.lower())

print("Cleaned Text:")
print(clean_text)
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize

sentences = sent_tokenize(text)
words = word_tokenize(clean_text)

print("Sentences:")
print(sentences)

print("\nWord Tokens:")
print(words)
```

```python
split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)

print("Using Python split():")
print(split_words)

print("\nUsing NLTK word_tokenize():")
print(nltk_words)

print("\nNumber of tokens using split():", len(split_words))
print("Number of tokens using word_tokenize():", len(nltk_words))
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in word_tokenize(clean_text)
    if word.isalpha() and word not in stop_words
]

print("Words after removing stopwords:")
print(filtered_words)
```

```python
from nltk.probability import FreqDist

frequency = FreqDist(filtered_words)

print("Word Frequencies:")
for word, count in frequency.most_common():
    print(word, ":", count)

frequency.plot(10, title="Top Word Frequencies")
plt.show()
```

Q3. Bag of Words and TF-IDF

```python
texts = [
    "Artificial intelligence helps computers learn and solve problems.",
    "Machine learning uses data to build intelligent computer systems.",
    "Online shopping makes product buying fast and convenient."
]

for i, text_item in enumerate(texts, start=1):
    print(f"Text {i}: {text_item}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

print("Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

feature_names = tfidf_vectorizer.get_feature_names_out()

print("TF-IDF Matrix:")
print(np.round(tfidf_matrix.toarray(), 3))
```

```python
for i, row in enumerate(tfidf_matrix.toarray(), start=1):
    top_indices = row.argsort()[-3:][::-1]

    print(f"\nText {i}:")
    for index in top_indices:
        print(feature_names[index], ":", round(row[index], 3))
```

Q4. Jaccard Similarity and Cosine Similarity

```python
ai_text = """
Artificial intelligence helps machines learn from data.
AI systems recognize patterns and solve complex problems.
They support automation, prediction, and decision making.
AI is used in healthcare, education, and transportation.
"""

blockchain_text = """
Blockchain stores information in linked digital blocks.
The technology supports secure and transparent transactions.
Blockchain networks can verify records without one central authority.
They are used in finance, contracts, and digital assets.
"""

print("Artificial Intelligence Text:")
print(ai_text)

print("\nBlockchain Text:")
print(blockchain_text)
```

```python
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = text.split()
    return tokens

ai_tokens = preprocess(ai_text)
blockchain_tokens = preprocess(blockchain_text)

print("AI Tokens:")
print(ai_tokens)

print("\nBlockchain Tokens:")
print(blockchain_tokens)
```

```python
# Convert token lists into sets
set_ai = set(ai_tokens)
set_blockchain = set(blockchain_tokens)

# Find intersection and union
intersection = set_ai.intersection(set_blockchain)
union = set_ai.union(set_blockchain)

# Calculate Jaccard similarity
jaccard_similarity = len(intersection) / len(union)

print("Common Words:", intersection)
print("Total Unique Words:", len(union))
print("Jaccard Similarity:", round(jaccard_similarity, 4))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

documents = [ai_text, blockchain_text]

vectorizer = TfidfVectorizer()
tfidf_vectors = vectorizer.fit_transform(documents)

cosine_score = cosine_similarity(
    tfidf_vectors[0:1],
    tfidf_vectors[1:2]
)[0][0]

print("Cosine Similarity:", round(cosine_score, 4))
```

```python
print("Jaccard Similarity:", round(jaccard_similarity, 4))
print("Cosine Similarity:", round(cosine_score, 4))

if cosine_score > jaccard_similarity:
    print("Cosine similarity gives a higher similarity score for these texts.")
elif jaccard_similarity > cosine_score:
    print("Jaccard similarity gives a higher similarity score for these texts.")
else:
    print("Both methods give the same similarity score.")
```

```python
def prepare_sequences_for_training(
    input_sequences: list,
    total_words: int
) -> tuple[np.ndarray, np.ndarray, int]:
    """
    Prepares input sequences for training by padding and splitting into features and labels.
    
    Args:
        input_sequences: List of tokenized sequences to be prepared for training.
        total_words: Total number of unique words in the vocabulary for one-hot encoding.
    
    Returns:
        tuple: A tuple containing:
            - X (np.ndarray): Feature sequences (all tokens except the last one).
            - y (np.ndarray): One-hot encoded labels (last token of each sequence).
            - max_sequence_len (int): Maximum sequence length after padding.
    
    Raises:
        ValueError: If input_sequences is empty.
    """
    if not input_sequences:
        raise ValueError("No input sequences were created. Add more training text.")
    
    max_sequence_len = max(len(seq) for seq in input_sequences)
    
    padded_sequences = np.array(
        pad_sequences(input_sequences, maxlen=max_sequence_len, padding="pre")
    )
    
    X = padded_sequences[:, :-1]
    y = padded_sequences[:, -1]
    y = tf.keras.utils.to_categorical(y, num_classes=total_words)
    
    print("Shape of X:", X.shape)
    print("Shape of y:", y.shape)
    print("Maximum sequence length:", max_sequence_len)
    
    return X, y, max_sequence_len
```

```python
def create_text_generation_model(
    total_words: int,
    embedding_dim: int = 16,
    lstm_units: int = 50
) -> Sequential:
    """
    Creates and compiles a sequential LSTM model for text generation.
    
    This model uses an embedding layer to convert word indices to dense vectors,
    an LSTM layer for sequence processing, and a dense output layer with softmax
    activation for word prediction.
    
    Args:
        total_words: The total number of unique words in the vocabulary.
        embedding_dim: Dimensionality of the embedding vectors. Default is 16.
        lstm_units: Number of LSTM units in the recurrent layer. Default is 50.
    
    Returns:
        A compiled Sequential model ready for training.
    """
    model = Sequential([
        Embedding(input_dim=total_words, output_dim=embedding_dim),
        LSTM(lstm_units),
        Dense(total_words, activation="softmax")
    ])
    
    model.compile(
        loss="categorical_crossentropy",
        optimizer="adam",
        metrics=["accuracy"]
    )
    
    return model

# Usage
model = create_text_generation_model(total_words=total_words)
model.summary()
```

```python
from typing import Any, Dict
import pandas as pd

def train_model(
    model: Any,
    X: pd.DataFrame,
    y: pd.Series,
    epochs: int = 50,
    verbose: int = 0
) -> Dict[str, list]:
    """
    Train a machine learning model on the provided dataset.
    
    This function trains the given model using the fit method and returns
    the training history containing metrics for each epoch.
    
    Args:
        model (Any): The machine learning model to train (e.g., Keras Sequential model).
            Must have a fit() method that returns a history object.
        X (pd.DataFrame): Training features as a pandas DataFrame.
        y (pd.Series): Training labels/targets as a pandas Series.
        epochs (int, optional): Number of training epochs. Defaults to 50.
        verbose (int, optional): Verbosity mode for training output:
            - 0: silent (no output)
            - 1: progress bar
            - 2: one line per epoch
            Defaults to 0.
    
    Returns:
        Dict[str, list]: Training history dictionary containing metrics 
            (e.g., loss, accuracy) for each epoch.
    
    Raises:
        AttributeError: If the model does not have a fit() method.
        ValueError: If X and y have incompatible shapes.
    
    Example:
        >>> history = train_model(model, X_train, y_train, epochs=100, verbose=1)
        >>> print(history.history.keys())
    """
    if not hasattr(model, 'fit'):
        raise AttributeError("The provided model must have a 'fit' method.")
    
    if len(X) != len(y):
        raise ValueError(f"X and y must have the same length. Got X: {len(X)}, y: {len(y)}")
    
    history = model.fit(
        X,
        y,
        epochs=epochs,
        verbose=verbose
    )
    
    print(f"Model training completed successfully! Trained for {epochs} epochs.")
    
    return history.history if hasattr(history, 'history') else history
```

```python
from typing import Optional
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

def generate_text(
    seed_text: str,
    next_words: int = 10,
    tokenizer = None,
    model = None,
    max_sequence_len: int = None
) -> str:
    """
    Generates text by predicting the next words based on a seed text.
    
    This function uses a trained language model to predict and append words
    sequentially to the seed text.
    
    Args:
        seed_text: The initial text to start generation from.
        next_words: Number of words to generate. Defaults to 10.
        tokenizer: Keras tokenizer object used for text-to-sequence conversion.
        model: Trained Keras model for word prediction.
        max_sequence_len: Maximum sequence length used during model training.
    
    Returns:
        The generated text string including the seed text and predicted words.
        
    Raises:
        ValueError: If tokenizer, model, or max_sequence_len is None.
    """
    if tokenizer is None or model is None or max_sequence_len is None:
        raise ValueError("tokenizer, model, and max_sequence_len must be provided")
    
    generated_text = seed_text.strip()

    for _ in range(next_words):
        # Convert text to token sequence
        token_list = tokenizer.texts_to_sequences([generated_text])[0]
        
        if not token_list:
            print(f"Warning: Seed word '{generated_text}' is not in the training vocabulary")
            break

        # Pad the sequence to match model input requirements
        padded_token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding="pre",
            truncating="pre"
        )

        # Predict next word probabilities
        predicted_probs = model.predict(padded_token_list, verbose=0)[0]
        predicted_word_index = int(np.argmax(predicted_probs))

        # Convert predicted index back to word
        output_word = tokenizer.index_word.get(predicted_word_index, "")
        
        if not output_word:
            print("Warning: Predicted word index not found in vocabulary")
            break

        generated_text += " " + output_word

    return generated_text
```
